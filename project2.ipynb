```python
# IMPORT LIBRARIES
#-------------------
import pandas as pd 
import numpy as np

# LOAD DATASET 
#-------------------
df =pd.read_csv(r"C:\Users\anshi\Downloads\student_performance_updated_1000.csv")

# UNDAERSTAND DATASET 
#--------------------
print(df.head())          #---> print starting 5 rows
print(df.shape)         #---> print (row,column)
print(df.columns)       #---> print column names 
print(df.info())          #---> print info about column and datatype

#  CHECKING MISSING VALUES 
#-------------------------------
print(df.isnull().sum())

#CHECKING DUPLICATES 
#----------------------------
print(df.duplicated().sum())


#FILLING MISSING VALUES 
#--------------------------
df.fillna({'AttendanceRate':df['AttendanceRate'].mean()},inplace=True)
df.fillna({'StudyHoursPerWeek':df['StudyHoursPerWeek'].mean()},inplace=True)
df.fillna({'PreviousGrade' :df['PreviousGrade'].mean()},inplace=True)
df.fillna({'ExtracurricularActivities':df['ExtracurricularActivities'].mean()},inplace=True)
df.fillna({'FinalGrade':df['FinalGrade'].mean()},inplace=True)
df.fillna({'Study Hours':df['Study Hours'].mean()},inplace=True)
df.fillna({'Attendance (%)': df['Attendance (%)'].mean()}, inplace=True)

# Removing remaining row with nan
#--------------------------------

df.dropna(inplace=True)


#Checking categorical value 
#---------------------------
print(df['Gender'].value_counts())
print(df['ParentalSupport'].value_counts())
print(df['Online Classes Taken'].value_counts())


# ============================================
# 10. CHECK NUMERICAL VALUES
# ============================================
print(
    df[
        [  'StudentID',                  
           'AttendanceRate',             
           'StudyHoursPerWeek',          
           'PreviousGrade',             
           'ExtracurricularActivities',  
           'FinalGrade',               
           'Study Hours',            
           'Attendance (%)',             
           'Online Classes Taken'       
        ]
        ].describe()

)
# ============================================
# 11. FINAL DATA VALIDATION
# ============================================

print("\n========== FINAL CHECK ==========")

# Final number of rows and columns
print("\nFinal shape:")
print(df.shape)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check duplicates
print("\nDuplicate rows:")
print(df.duplicated().sum())

# Check data types
print("\nData types:")
print(df.dtypes)


# ============================================
# 12. SAVE CLEANED DATASET
# ============================================

df.to_csv(
    r"C:\Users\anshi\Downloads\Cleaned_student_performance_updated.csv",
    index=False
)

print("\nData cleaning completed successfully!")
print("Cleaned dataset saved as 'Cleaned_student_performance_updated.csv'")
```
